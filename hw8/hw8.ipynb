#Работа с API

```python
import requests
import pandas as pd
```

```python
base_url = "https://anapioficeandfire.com/api"
```

Получить информацию обо всех книгах и сохранить в pd.DataFrame

```python
books = requests.get(f"{base_url}/books").json()
```

```python
df_books = pd.DataFrame(books)
display(df_books)
```

Получить информацию обо всех домах Вестероса (Houses) и сохранить в pd.DataFrame

```python
houses = requests.get(f"{base_url}/houses?pageSize=50").json()

df_houses = pd.DataFrame(houses)
display(df_houses)
```

Получить информацию обо всех домах Вестероса, у которых есть девиз, и сохранить в pd.DataFrame

```python
houses_with_titles = requests.get(f"{base_url}/houses?pageSize=50&hasTitles=true").json()
df_houses_with_titles = pd.DataFrame(houses_with_titles)
display(df_houses_with_titles[["name", "titles"]].head())
```
