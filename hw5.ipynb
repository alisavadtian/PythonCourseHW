#Polars

```python
import polars as pl

df = pl.read_csv("/content/train (1).csv")
print(df.head())
```

Выведите основную информацию о датасете

```python
print("Data Types:")
print(df.dtypes)
print("\nNull Counts:")
print(df.null_count())
print("\nDescriptive Statistics:")
print(df.describe())
print("Mean:")
```

Посчитайте количество пассажиров каждого класса

```python
pclass_counts = df.get_column("Pclass").value_counts()
print(pclass_counts)
```

Выведите количество выживших мужчин и женщин на корабле

```python
survived_by_sex = df.group_by("Sex").agg(pl.col("Survived").sum().alias("SurvivedCount"))
print(survived_by_sex)
```

Выведите часть таблицы с пассажирами, возраст которых больше 44 лет

```python
older_passengers = df.filter(pl.col("Age") > 44)
print(older_passengers)
```

```python
import polars as pl

df = pl.read_csv("/content/train (1).csv")
print(df.head())
```

Посчитайте средний возраст пассажиров и его стандартное отклонение с помощью bottleneck

```python
import bottleneck as bn

mean_age = bn.nanmean(df.get_column("Age"))
std_age = bn.nanstd(df.get_column("Age"))

print(f"Средний возраст пассажиров: {mean_age:.2f}")
print(f"Стандартное отклонение возраста пассажиров: {std_age:.2f}")
```

```python
fare_new_list = [(row['Fare'] * 1.3) for row in df.iter_rows(named=True)]
df = df.with_columns(pl.Series("Fare_new", fare_new_list))

print(df.head())
```

#Оптимизация типов pandas

```python
import pandas as pd

df_housing = pd.read_csv("/content/Housing.csv")
print(df_housing.head())
```

```python
print(df_housing.dtypes)


# 'price' int64 to int32
# 'area' int64 to int32.
# 'bedrooms' int64 to int8.
# 'bathrooms' int64 to  int8.
# 'stories' int64. Can likely be downcasted to a smaller integer type like int8.
# 'mainroad' object to 'category' (yes/no).
# 'guestroom' object to 'category'.
# 'basement' object to 'category'.
# 'hotwaterheating' object to 'category'.
# 'airconditioning' object 'category'.
# 'parking' int64 to int8.
# 'prefarea' object to 'category'.
# 'furnishingstatus' object to 'category'
```

```python
print("Memory usage before optimization:")
print(df_housing.memory_usage(deep=True))


df_housing['price'] = df_housing['price'].astype('int32')
df_housing['area'] = df_housing['area'].astype('int32')
df_housing['bedrooms'] = df_housing['bedrooms'].astype('int8')
df_housing['bathrooms'] = df_housing['bathrooms'].astype('int8')
df_housing['stories'] = df_housing['stories'].astype('int8')
df_housing['mainroad'] = df_housing['mainroad'].astype('category')
df_housing['guestroom'] = df_housing['guestroom'].astype('category')
df_housing['basement'] = df_housing['basement'].astype('category')
df_housing['hotwaterheating'] = df_housing['hotwaterheating'].astype('category')
df_housing['airconditioning'] = df_housing['airconditioning'].astype('category')
df_housing['parking'] = df_housing['parking'].astype('int8')
df_housing['prefarea'] = df_housing['prefarea'].astype('category')
df_housing['furnishingstatus'] = df_housing['furnishingstatus'].astype('category')

print("\nMemory usage after optimization:")
print(df_housing.memory_usage(deep=True))
print("\nOptimized Data Types:")
print(df_housing.dtypes)
```
